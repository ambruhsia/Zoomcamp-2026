### Q3. Fuel types

```python
# Q3. Fuel types
num_fuel_types = df['fuel_type'].nunique()
print(f"Number of unique fuel types: {num_fuel_types}")
```

### Q4. Missing values

```python
# Q4. Missing values
missing_values_count = df.isnull().sum()
columns_with_missing_values = missing_values_count[missing_values_count > 0].index.tolist()
num_columns_with_missing_values = len(columns_with_missing_values)
print(f"Number of columns with missing values: {num_columns_with_missing_values}")
print(f"Columns with missing values: {columns_with_missing_values}")
```

### Q5. Max fuel efficiency

```python
# Q5. Max fuel efficiency of cars from Asia
asia_cars = df[df['origin'] == 'Asia']
max_fuel_efficiency_asia = asia_cars['fuel_efficiency_mpg'].max()
print(f"Maximum fuel efficiency of cars from Asia: {max_fuel_efficiency_asia}")
```

### Q6. Median value of horsepower

```python
# Q6. Median value of horsepower

# Initial median horsepower
median_horsepower_initial = df['horsepower'].median()
print(f"Initial median horsepower: {median_horsepower_initial}")

# Most frequent value of horsepower
most_frequent_horsepower = df['horsepower'].mode()[0]
print(f"Most frequent horsepower: {most_frequent_horsepower}")

# Fill missing values in 'horsepower' with the most frequent value
df_filled = df.copy()
df_filled['horsepower'].fillna(most_frequent_horsepower, inplace=True)

# New median horsepower after filling missing values
median_horsepower_after_fill = df_filled['horsepower'].median()
print(f"Median horsepower after filling missing values: {median_horsepower_after_fill}")

# Has it changed?
if median_horsepower_after_fill > median_horsepower_initial:
    print("Median has increased.")
elif median_horsepower_after_fill < median_horsepower_initial:
    print("Median has decreased.")
else:
    print("Median has not changed.")
```

### Q7. Sum of weights (Linear Regression Implementation)

```python
# Q7. Sum of weights

# Select all cars from Asia
asia_cars_q7 = df[df['origin'] == 'Asia']

# Select only columns vehicle_weight and model_year
asia_cars_q7_selected = asia_cars_q7[['vehicle_weight', 'model_year']]

# Select the first 7 values
asia_cars_q7_first_7 = asia_cars_q7_selected.head(7)

# Get the underlying NumPy array. Let's call it X.
X = asia_cars_q7_first_7.values
print("X array:\n", X)

# Compute matrix-matrix multiplication between the transpose of X and X. Let's call the result XTX.
XTX = X.T @ X
print("XTX matrix:\n", XTX)

# Invert XTX.
XTX_inv = np.linalg.inv(XTX)
print("Inverse of XTX:\n", XTX_inv)

# Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print("y array:\n", y)

# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = XTX_inv @ X.T @ y
print("w array (coefficients):\n", w)

# What's the sum of all the elements of the result?
sum_of_w = w.sum()
print(f"Sum of all elements of w: {sum_of_w:.4f}")
```
